### ЗАДАЧА: Реестр товаров 

В этой задаче нужно собрать простой каталог товаров для внутренней системы закупок.
Данные приходят как строки, часть строк может быть некорректной.

НЕОБХОДИМО РЕАЛИЗОВАТЬ:

1. Класс `Product`:
   - поля `id`, `name`, `_price`, `category`
   - `@property price` (только чтение)
   - метод `set_price(value) -> bool`:
     - возвращает `True`, если цена валидна и обновлена
     - возвращает `False`, если цена невалидна
   - `@classmethod from_line(cls, raw)`:
     - если строка корректная, вернуть `(product, "")`
     - если ошибка, вернуть `(None, "текст причины")`.

2. Класс `ProductRegistry`:
   - хранит список товаров
   - метод `add(product) -> (bool, str)`:
     - `(True, "")`, если успешно
     - `(False, "...")`, если `id` уже существует
   - метод `count()`
   - метод `all_products()`
   - метод `has_id(id)`
   - метод `by_category()` -> `{category: count}`
   - метод `avg_price()`.

3. Загрузка данных:
   - пройти по `lines`
   - использовать `Product.from_line(...)`
   - если товар создан, пытаться `registry.add(...)`
   - все проблемные строки складывать в `problems` как `(line, reason)`.

4. Вывод:
   - число валидных товаров
   - статистика по категориям
   - средняя цена
   - список проблемных строк.


- Для проверки дублей хранить set из `id`.
- В `avg_price()` вернуть `0`, если товаров нет.
- Для проверки, что строка число, можно использовать вспомогательную функцию с `str.replace('.', '', 1).isdigit()`.


```python
def is_number(text):
    try:
        float(text)
        return True
    except ValueError:
        return False


class Product:
    def __init__(self, id_, name, price, category):
        self.id = id_
        self.name = name
        self._price = price
        self.category = category

    @property
    def price(self):
        return self._price

    def set_price(self, value):
        if isinstance(value, (int, float)) and value >= 0:
            self._price = value
            return True
        return False

    @classmethod
    def from_line(cls, raw):
        
        parts = raw.split(";")
        if len(parts) != 4:
            return None, "Неверное количество полей"
        id_, name, price_str, category = parts

        if not is_number(price_str):
            return None, f"Цена не является числом: {price_str}"

        price = float(price_str)
        if price < 0:
            return None, f"Цена не может быть отрицательной: {price}"

        return cls(id_, name, price, category), None


class ProductRegistry:
    def __init__(self):
        self.products = []
        self.ids = set()

    def add(self, product):
        if product.id in self.ids:
            return False, "Дубликат ID"
        self.products.append(product)
        self.ids.add(product.id)
        return True, None

    def count(self):
        return len(self.products)

    def all_products(self):
        return self.products.copy()

    def has_id(self, id_):
        return id_ in self.ids

    def by_category(self):
        categories = {}
        for prod in self.products:
            categories.setdefault(prod.category, []).append(prod)
        return categories

    def avg_price(self):
        if not self.products:
            return 0.0
        total = sum(prod.price for prod in self.products)
        return total / len(self.products)



registry = ProductRegistry()
problems = []

for line in lines:
    product, reason = Product.from_line(line)
    if product is None:
        problems.append((line, reason))
    else:
        ok, reason_add = registry.add(product)
        if not ok:
            problems.append((line, reason_add))

print(f"Всего товаров: {registry.count()}")

print("Товары по категориям:")
by_cat = registry.by_category()
for cat, prods in by_cat.items():
    print(f"{cat}:")
    for p in prods:
        print(f"  {p.id} - {p.name} - {p.price}")

print(f"Средняя цена товаров: {registry.avg_price():.2f}")

if problems:
    print("Проблемные записи:")
    for line, reason in problems:
        print(f"{line} -> {reason}")
else:
    print("Проблем не обнаружено.")
```
