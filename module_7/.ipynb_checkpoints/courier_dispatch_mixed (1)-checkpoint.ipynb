### ЗАДАЧА: Распределение доставок между курьерами

Логистическая команда получает пакет заявок на доставки, которые нужно распределить между курьерами на электровелосипедах.
Нужно собрать систему, которая:
- принимает корректные доставки,
- отбрасывает неправильные или небезопасные заявки,
- уменьшает доступный заряд после успешно назначенного маршрута,
- ведёт отдельный журнал ошибок,
- помогает понять, какой курьер был загружен дольше всех и какому клиенту доставили самый большой вес.

```python
from dataclasses import dataclass
from typing import Optional


couriers_data = {
    'CR-1': {'zone': 'north', 'charge_min': 40, 'max_weight': 3.0},
    'CR-2': {'zone': 'south', 'charge_min': 30, 'max_weight': 2.0},
    'CR-3': {'zone': 'north', 'charge_min': 55, 'max_weight': 5.0},
}

# rows: delivery_id|courier_id|client|weight_kg|route_min
rows = [
    'DL-100|CR-1|Clinic|1.5|12',
    'DL-101|CR-2|Cafe|2.5|10',
    'DL-102|CR-9|Lab|1.0|8',  # Invalid courier
    'DL-103|CR-1|Shop|0|6',  # Invalid weight
    'DL-104|CR-3|Village|3.5|60', # Exceeds max weight
    'DL-100|CR-3|Clinic|1.0|10', # Duplicate delivery ID
    'DL-105|CR-3|School|2.0|20',
    'DL-106|CR-2|Pharmacy|1.0|15',
]


class DeliveryError(Exception):
    pass


class RowFormatError(DeliveryError):
    pass


class CourierNotFoundError(DeliveryError):
    pass


class WeightError(DeliveryError):
    pass


class RouteTimeError(DeliveryError):
    pass


class WeightLimitError(DeliveryError):
    pass


class ChargeReserveError(DeliveryError):
    pass


class DuplicateDeliveryError(DeliveryError):
    pass


@dataclass(order=True)
class Delivery:
    route_min: int
    delivery_id: str
    courier_id: str
    client: str
    weight_kg: float


class Courier:
    def __init__(self, courier_id: str, zone: str, charge_min: int, max_weight: float):
        self.courier_id = courier_id
        self.zone = zone
        self.charge_min = charge_min
        self.max_weight = max_weight
        self.deliveries: List[Delivery] = []
        self._charge_left = charge_min  # Internal state for charge

    def charge_left(self) -> int:
        return self._charge_left

    def total_route_time(self) -> int:
        return sum(delivery.route_min for delivery in self.deliveries)

    def total_weight(self) -> float:
        return sum(delivery.weight_kg for delivery in self.deliveries)

    def assign(self, delivery: Delivery):
        if delivery.weight_kg > self.max_weight:
            raise WeightLimitError(f"Delivery {delivery.delivery_id} exceeds max weight for courier {self.courier_id} ({delivery.weight_kg}kg > {self.max_weight}kg)")

        charge_after_deduction = self._charge_left - delivery.route_min
        if charge_after_deduction < 5:
            raise ChargeReserveError(f"Deducting route time for {delivery.delivery_id} would leave courier {self.courier_id} with insufficient charge ({charge_after_deduction} < 5)")

        self.deliveries.append(delivery)
        self.deliveries.sort()  # Sort by route_min (due to dataclass order=True)
        self._charge_left = charge_after_deduction

    def __repr__(self):
        return (f"Courier(id='{self.courier_id}', zone='{self.zone}', "
                f"charge_left={self.charge_left()}, deliveries={len(self.deliveries)})")


class CourierDispatchServicecouriers[cid] = Courier(cid, data['zone'], data['charge_min'], data['max_weight'])

        self.accepted: List[Delivery] = []
        self.errors: List[Tuple[str, str, str]] = []
        self.processed_ids: set[str] = set()

    def parse_delivery(self, row: str) -> Delivery:
        parts = [p.strip() for p in row.split('|')]
        if len(parts) != 5:
            raise RowFormatError(f"Expected 5 parts, but got {len(parts)} in row: '{row}'")

        delivery_id, courier_id, client, weight_raw, route_raw = parts

        if courier_id not in self.couriers:
            raise CourierNotFoundError(f"Courier ID '{courier_id}' not found in row: '{row}'")

        try:
            weight_kg = float(weight_raw)
            if weight_kg <= 0:
                raise WeightError(f"Weight must be positive, got {weight_kg}kg in row: '{row}'")
        except ValueError as exc:
            raise WeightError(f"Invalid weight format '{weight_raw}' in row: '{row}'") from exc

        try:
            route_min = int(route_raw)
            if route_min <= 0:
                raise RouteTimeError(f"Route time must be positive, got {route_min} minutes in row: '{row}'")
        except ValueError as exc:
            raise RouteTimeError(f"Invalid route time format '{route_raw}' in row: '{row}'") from exc

        return Delivery(
            delivery_id=delivery_id,
            courier_id=courier_id,
            client=client,
            weight_kg=weight_kg,
            route_min=route_min
        )

    def submit(self, row: str):
        try:
            delivery = self.parse_delivery(row)

            if delivery.delivery_id in self.processed_ids:
                raise DuplicateDeliveryError(f"Duplicate delivery ID '{delivery.delivery_id}' in row: '{row}'")

            courier = self.couriers[delivery.courier_id]
            courier.assign(delivery)

            self.processed_ids.add(delivery.delivery_id)
            self.accepted.append(delivery)

        except DeliveryError as e:
            self.errors.append((row, type(e).__name__, str(e)))

    def load(self, rows: List[str]):
        for row in rows:
            self.submit(row)

    def client_weights(self) -> Dict[str, float]:
        client_weights_map: Dict[str, float] = {}
        for delivery in self.accepted:
            client_weights_map[delivery.client] = client_weights_map.get(delivery.client, 0) + delivery.weight_kg
        return client_weights_map

    def top_client(self) -> Optional[Tuple[str, float]]:
        weights = self.client_weights()
        if not weights:
            return None
        top_client = max(weights.items(), key=lambda item: item[1])
        return top_client

    def busiest_courier(self) -> Optional[Tuple[str, int]]:
        if not self.couriers:
            return None

        busiest_courier_id = None
        max_route_time = -1

        for courier_id, courier in self.couriers.items():
            current_route_time = courier.total_route_time()
            if current_route_time > max_route_time:
                max_route_time = current_route_time
                busiest_courier_id = courier_id

        if busiest_courier_id is not None:
            return busiest_courier_id, max_route_time
        return None

    def low_charge_couriers(self, threshold: int = 15) -> List[Tuple[str, int]]:
        low_charge = []
        for courier_id, courier in self.couriers.items():
            if courier.charge_left() <= threshold:
                low_charge.append((courier_id, courier.charge_left()))
        return low_charge

    def find_delivery(self, delivery_id: str) -> Optional[Delivery]:
        for courier in self.couriers.values():
            for delivery in courier.deliveries:
                if delivery.delivery_id == delivery_id:
                    return delivery
        return None


service = CourierDispatchService(couriers_data)

# TODO: загрузить rows через service.load(rows)
service.load(rows)

# TODO: вывести принятые доставки
print("--- Accepted Deliveries ---")
if not service.accepted:
    print("No deliveries accepted.")
else:
    for delivery in service.accepted:
        print(f"  {delivery}")

# TODO: вывести ошибки
print("\n--- Errors ---")
if not service.errors:
    print("No errors encountered.")
else:
    for row, error_type, message in service.errors:
        print(f"  Row: '{row}' | Error Type: {error_type} | Message: {message}")

# TODO: вывести по каждому курьеру deliveries, total_route_time и charge_left
print("\n--- Courier Status ---")
for courier_id, courier in service.couriers.items():
    print(f"  Courier ID: {courier_id}")
    print(f"    Zone: {courier.zone}")
    print(f"    Max Weight: {courier.max_weight}kg")
    print(f"    Number of Deliveries: {len(courier.deliveries)}")
    print(f"    Total Route Time: {courier.total_route_time()} minutes")
    print(f"    Charge Left: {courier.charge_left()} minutes")
    if courier.deliveries:
        print("    Deliveries assigned (sorted by route time):")
        for delivery in courier.deliveries:
            print(f"      - {delivery}")
    print("-" * 20)

# TODO: вывести top_client()
top_client_info = service.top_client()
print("\n--- Top Client ---")
if top_client_info:
    client, weight = top_client_info
    print(f"  Client: {client}, Total Weight: {weight:.2f}kg")
else:
    print("  No clients found.")

# TODO: вывести busiest_courier()
busiest_courier_info = service.busiest_courier()
print("\n--- Busiest Courier ---")
if busiest_courier_info:
    courier_id, route_time = busiest_courier_info
    print(f"  Courier ID: {courier_id}, Total Route Time: {route_time} minutes")
else:
    print("  No couriers found.")

# TODO: вывести low_charge_couriers()
low_charge_couriers_list = service.low_charge_couriers()
print("\n--- Couriers with Low Charge (<= 15 minutes left) ---")
if not low_charge_couriers_list:
    print("  All couriers have sufficient charge.")
else:
    for courier_id, charge in low_charge_couriers_list:
        print(f"  Courier ID: {courier_id}, Charge Left: {charge} minutes")

# TODO: вывести find_delivery('DL-105')
delivery_105 = service.find_delivery('DL-105')
print("\n--- Finding Delivery DL-105 ---")
if delivery_105:
    print(f"  Found: {delivery_105}")
else:
    print("  Delivery DL-105 not found.")

delivery_nonexistent = service.find_delivery('DL-999')
print("\n--- Finding Delivery DL-999 ---")
if delivery_nonexistent:
    print(f"  Found: {delivery_nonexistent}")
else:
    print("  Delivery DL-999 not found.")
```
